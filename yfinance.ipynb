```python
import yfinance as yf
import pandas as pd
```

```python
tesla = yf.Ticker("TSLA")
```

```python
tesla_data = tesla.history(period="1y")
```

```python
print(tesla_data.head())
```

```python


import requests
from bs4 import BeautifulSoup

```

```python
url = "https://www.macrotrends.net/stocks/charts/TSLA/tesla/revenue"
html_data = requests.get(url).text

soup = BeautifulSoup(html_data, "html5lib")

tesla_revenue = pd.DataFrame(columns=["Date", "Revenue"])

# Find all tables on the page
tables = soup.find_all('table')

# Iterate through the tables and find the one with revenue data
for table in tables:
    if 'Tesla Quarterly Revenue' in table.text:  # Look for a table containing this text
        for row in table.find("tbody").find_all("tr"):
            col = row.find_all("td")
            if col: # Check if col is not empty
                date = col[0].text
                revenue = col[1].text.replace("$", "").replace(",", "")

                if date and revenue:
                    tesla_revenue = tesla_revenue.append({"Date": date, "Revenue": revenue}, ignore_index=True)
        break # Stop after finding the correct table

tesla_revenue["Revenue"] = tesla_revenue["Revenue"].astype(float)

print(tesla_revenue.tail())
```

```python

gamestop = yf.Ticker("GME")

gamestop_data = gamestop.history(period="max")
print(gamestop_data.head())
```

```python


url = "https://www.macrotrends.net/stocks/charts/GME/gamestop/revenue"
html_data = requests.get(url).text

soup = BeautifulSoup(html_data, "html5lib")

gamestop_revenue = pd.DataFrame(columns=["Date", "Revenue"])

tables = soup.find_all('table')

for table in tables:
    if 'GameStop Quarterly Revenue' in table.text:
        for row in table.find("tbody").find_all("tr"):
            col = row.find_all("td")
            if col:
                date = col[0].text
                revenue = col[1].text.replace("$", "").replace(",", "")
                if date and revenue:
                     gamestop_revenue = gamestop_revenue.append({"Date": date, "Revenue": revenue}, ignore_index=True)
        break

gamestop_revenue["Revenue"] = gamestop_revenue["Revenue"].astype(float)

print(gamestop_revenue.head())
```

```python

import plotly.graph_objects as go
from plotly.subplots import make_subplots

def make_graph(stock_data, revenue_data, stock):
  fig = make_subplots(rows=2, cols=1, shared_xaxes=True, subplot_titles=("Historical Share Price", "Historical Revenue"), vertical_spacing = .3)
  fig.add_trace(go.Scatter(x=pd.to_datetime(stock_data.index, infer_datetime_format=True), y=stock_data.Close.astype("float"), name="Share Price"), row=1, col=1)
  fig.add_trace(go.Scatter(x=pd.to_datetime(revenue_data.Date, infer_datetime_format=True), y=revenue_data.Revenue.astype("float"), name="Revenue"), row=2, col=1)
  fig.update_xaxes(title_text="Date", rangeslider_visible=True, row=1, col=1)
  fig.update_xaxes(title_text="Date", rangeslider_visible=True, row=2, col=1)
  fig.update_yaxes(title_text="Price ($)", row=1, col=1)
  fig.update_yaxes(title_text="Revenue ($ Millions)", row=2, col=1)
  fig.update_layout(showlegend=False,
  height=600,
  title=stock,
  xaxis1_rangeslider_visible=True,
  xaxis2_rangeslider_visible=True)
  fig.show()

make_graph(tesla_data, tesla_revenue, 'Tesla Stock Data and Revenue')
```

```python


make_graph(gamestop_data, gamestop_revenue, 'GameStop Stock Data and Revenue')
```
